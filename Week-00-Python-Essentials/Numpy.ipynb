### **Why is NumPy faster than standard Python lists?**

* Fixed Data Types and Smaller Memor - NumPy uses fixed types (such as int32, int16, or int8), storing data compactly. Standard Python lists store full built-in int objects that require additional memory to track object values, object types, reference counts, and size. Because NumPy uses far fewer bytes of memory, the computer reads and processes the data much quicke
* No Type Checking During Iteration
